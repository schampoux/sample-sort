```python
import logging 
from pathlib import Path
import os 
import yaml 
import re 
from typing import List 

def classify(path: Path, rules) -> str:
    for pattern, category in rules:
        if pattern.search(path):
            return category 
    return "Uncategorized"

def load_rules(path='./rules.yaml'):
    with open(path, 'r') as f: 
        config=yaml.safe_load(f)
    compiled_rules=[
        (re.compile(rule['pattern']), rule['category'])
        for rule in config['rules']
    ]
    return compiled_rules

def fill_path(path: str):
    path.replace(' ', '-')
    return path 

packs_path = '/home/spunion/Documents/splice' # "sounds/packs"
base_path = '/home/spunion/Documents/splice' 
output_path = '/home/spunion/Documents/splice/sorted-sounds'

packs = os.listdir(packs_path)

def classify_packs() -> List[str]:
    unclassified_packs = []
    classified_file_paths = []
    for sound_pack in packs:

        sound_pack_path = os.path.join(base_path, packs_path, sound_pack)

        for dirpath, dirnames, filenames in os.walk(sound_pack_path):
            if filenames:
                # remove desktop services store (mac metadata file)
                if ".DS_Store" in filenames:
                    if len(filenames) == 1:
                        filenames.remove(".DS_Store")
                        print(f"Directory {dirpath} does not contain any sound files.")
                        print("\nNEXT\n")
                        continue
                    else:
                        filenames.remove(".DS_Store") 

                print("\npack: ", sound_pack, "\ndirpath: ", dirpath, "\ndirnames: ", dirnames, "\nfilenames: ", filenames)
                classification_dir = classify(path = dirpath, rules = load_rules())     

                if classification_dir == "Uncategorized":
                    print(f"Current directory {dirpath} Uncategorized, cannot parse directory name for naming convention.")
                    for file in filenames: 
                        unclassified_packs.append(os.path.join(dirpath, file))
                    print("\nNEXT\n")
                    continue

                for file_name in filenames: 
                    print("split dirpath: ",dirpath.split('/'))
                    print("file_name: ", file_name)
                    classification_file = classify(path = os.path.join(dirpath[-1], file_name), rules = load_rules())
                    classified_file_paths.append(classification_file)
                    print("dir class: ", classification_dir, "\nfile class: ",classification_file, "\n")
        
    return unclassified_packs, classification_file
unclassified_packs = classify_packs()
unclassified_packs
```

# Regex

`.*` match anything before this, including nothing. 

    - `.` match anything 
    - `*` zero or more times
